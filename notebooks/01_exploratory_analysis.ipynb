This notebook explores the cleaned VPD crime dataset
(`data/processed/crime_data_cleaned.csv`) produced by the pipeline in
`scripts/`. Findings here feed the Streamlit dashboard, the statistical
analysis, and the machine-learning phase.

> Reminder: the dataset contains *reported, founded* incidents only, is
> missing 2022 entirely, and locations are anonymized to hundred-block
> level. Do not over-interpret patterns as causes.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

CLEAN = Path("../data/processed/crime_data_cleaned.csv")
df = pd.read_csv(CLEAN, parse_dates=["INCIDENT_DATETIME"])
df.head()
```

```python
print(f"Rows: {len(df):,}")
print(f"Columns: {len(df.columns)}")
print(df.dtypes)
```

```python
missing = df.isna().sum()
missing[missing > 0].sort_values(ascending=False)
```

## Crime type distribution

```python
type_counts = df["TYPE"].value_counts()
type_counts
```

```python
type_counts.plot(kind="barh", figsize=(8, 6), title="Incidents by crime type")
plt.xlabel("Incidents")
plt.tight_layout()
plt.show()
```

## Yearly trend

```python
by_year = df.groupby("YEAR").size()
by_year.plot(kind="line", marker="o", figsize=(10, 4), title="Incidents per year")
plt.ylabel("Incidents")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## Neighbourhood distribution

```python
hood_counts = df["NEIGHBOURHOOD"].value_counts()
hood_counts.head(10)
```

```python
hood_counts.head(10).plot(kind="barh", figsize=(8, 6), title="Top 10 neighbourhoods")
plt.xlabel("Incidents")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

## Temporal patterns

```python
df["HOUR"].value_counts().sort_index().plot(
    kind="bar", figsize=(10, 4), title="Incidents by hour of day"
)
plt.xlabel("Hour (0-23)")
plt.ylabel("Incidents")
plt.tight_layout()
plt.show()
```

```python
day_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
df["DAY_OF_WEEK"].value_counts().reindex(day_order).plot(
    kind="bar", figsize=(8, 4), title="Incidents by day of week"
)
plt.ylabel("Incidents")
plt.tight_layout()
plt.show()
```

```python
season_order = ["Winter", "Spring", "Summer", "Fall"]
df["SEASON"].value_counts().reindex(season_order).plot(
    kind="bar", figsize=(6, 4), title="Incidents by season"
)
plt.ylabel("Incidents")
plt.tight_layout()
plt.show()
```

## Check the 2022 gap in the year x month grid

## Key findings (to be refined in the statistical phase)

- 11 crime types; "Theft from Vehicle" and "Mischief" are usually the
  most frequent categories in VPD data (confirm in the current download).
- Clear weekday/weekend and day/night structure expected; peak hours are
  typically afternoon-evening.
- 2022 is entirely absent - keep flagged for the final report.
- 24 neighbourhoods; the Downtown area dominates counts (population and
  foot-traffic density), which matters when comparing rates rather than
  raw counts.
