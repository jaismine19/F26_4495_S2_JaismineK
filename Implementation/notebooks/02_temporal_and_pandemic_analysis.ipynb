## Seasonality checkHypothesis: Theft from Vehicle is worse in summer (more cars parked,windows down). Test it against the all-crime seasonal baseline.

```python
tfv = df[df["TYPE"] == "Theft from Vehicle"]season_order = ["Winter", "Spring", "Summer", "Fall"]s = tfv["SEASON"].value_counts().reindex(season_order)print("Theft from Vehicle by season:")for season in season_order:    print(f"  {season:8s} {s[season]:,}  ({s[season]/len(tfv)*100:.1f}%)")base = df["SEASON"].value_counts().reindex(season_order) / len(df) * 100print("\nAll crime baseline (%):")print(base.round(1))
```

**Finding 4 - The hypothesis that failed.** Theft from Vehicle is flatacross seasons (24.2 / 24.7 / 25.3 / 25.7%) even though all crime has asmall summer bump (26.5% vs 23.6% in winter). The hypothesis is rejected -a rejected hypothesis is still a result worth reporting.

## Neighbourhood changeHow did incident counts change per neighbourhood between 2019 and 2023?

```python
hood = (df[df["YEAR"].isin([2019, 2023])]         .groupby(["NEIGHBOURHOOD", "YEAR"])         .size()         .unstack())hood["change_pct"] = (hood[2023] - hood[2019]) / hood[2019] * 100print("Top 5 improvers (% change 2019 -> 2023):")print(hood.nsmallest(5, "change_pct")[["change_pct"]].round(1))print("\nNeighbourhoods with increases:")print(hood[hood["change_pct"] > 0][["change_pct"]].round(1))
```

**Finding 5 - Neighbourhoods diverge.** 21 of 24 neighbourhoods hadfewer incidents in 2023 than 2019. The biggest improvers were Oakridge(-50.7%), Riley Park (-46.2%), and Hastings-Sunrise (-42.7%). Only Musqueam(+25.8%), Renfrew-Collingwood (+21.3%), and Sunset (+10.8%) increased.Caveats: Musqueam is small, so its percentages are noisy; 2022 is missing;and population shifts can explain part of these changes.
